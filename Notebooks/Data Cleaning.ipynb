#### Importing data using pandas

```python
import pandas as pd
df=pd.read_csv("customer_shopping_behavior.csv")
```

```python
## insulting null value of review rating with the median of each category
df['Review Rating']=df.groupby('Category')['Review Rating'].transform(lambda x:x.fillna(x.median()))
```

```python
df.isnull().sum()
```

```python
## Rename columns according to snake casing for better readability and documentation
df.columns=df.columns.str.lower()
df.columns=df.columns.str.replace(' ','_')
df=df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
```

```python
df.columns
```

```python
## creating a new column to classify age into age groups

labels=['Young Adult','Adult','Middle-aged','Senior']
df['age_group']=pd.qcut(df['age'],q=4,labels=labels)
```

```python
df[['age','age_group']].head(10)
```

```python
df['frequency_of_purchases'].unique()
```

```python
## create column purchase_frequency_days
frequency_mapping={
    'Fortnightly':14,
    'Weekly':7,
    'Monthly':30,
    'Quarterly':90,
    'Bi-Weekly':14,
    'Every 3 Months':90,
    'Annually':365
}

df['purchase_frequency_days']=df['frequency_of_purchases'].map(frequency_mapping)
```

```python
df[['frequency_of_purchases','purchase_frequency_days']].head()
```

```python
df['purchase_frequency_days'].isnull().sum()## checking null value
```

```python
df[['discount_applied','promo_code_used']]
```

```python
## drop promo code use

df=df.drop('promo_code_used',axis=1)
```
